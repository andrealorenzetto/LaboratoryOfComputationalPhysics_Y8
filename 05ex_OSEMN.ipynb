2\. Create a random matrix of 5x5 and then save it to a text file named "data.txt"

```python
m = np.random.randint(0,10,25)
m = m.reshape(5,5)
print(m)
with open('C:/Users/Utente/Desktop/test/data.txt', 'w') as f:
    for i in range(5):
        for j in range(5):
            f.write(str(m[i][j]) + '\t')
        f.write('\n')
```

3\. Load the saved txt file of point 2 and convert it to a csv file (by hand)

```python
with open('C:/Users/Utente/Desktop/test/data.txt', 'r') as f:
    with open('C:/Users/Utente/Desktop/test/data.csv', 'w') as f1:
        for line in f:
            index = 0
            for c in line.split():
                if(index == 0):
                    f1.write(c)
                else:
                    f1.write(','+c)
                index += 1
            f1.write('\n')
```
