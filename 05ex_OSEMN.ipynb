3\. Load the saved txt file of point 2 and convert it to a csv file (by hand)

```python
with open('C:/Users/Utente/Desktop/test/data.txt', 'r') as f:
    with open('C:/Users/Utente/Desktop/test/data.csv', 'w') as f1:
        for line in f:
            index = 0
            for c in line.split():
                if(index == 0):
                    f1.write(c)
                else:
                    f1.write(','+c)
                index += 1
            f1.write('\n')
```

7\. Load the remote file [https://www.dropbox.com/s/vkl89yce7xjdq4n/regression_generated.csv?dl=1](https://www.dropbox.com/s/vkl89yce7xjdq4n/regression_generated.csv?dl=1) with Pandas and plot a scatter plot all possible combination of the following fields:
    
  + features_1
  + features_2
  + features_3

```python
file_name = 'C:/Users/Utente/Downloads/regression_generated.csv'
df=pd.read_csv(file_name, skiprows=0)
fig, ax = plt.subplots(nrows = 3, ncols = 3, figsize = (20,20))
features = ['features_' + str(i+1) for i in range(3)]
i, j = 0, 0
for f1 in features:
    j = 0
    for f2 in features:
        ax[i][j].scatter(df[f1],df[f2], s = 0.7)
        ax[i][j].set_title(str(i+1) +' vs '+str(j+1))
        j += 1
    i += 1
plt.show()
```
